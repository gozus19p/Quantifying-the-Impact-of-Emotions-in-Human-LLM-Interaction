```python
import matplotlib.pyplot as plt
import pandas as pd

# Data from the table
data = {
    "Emotion": ["Joy", "Anger", "Fear"],
    "Baseline": [6, 0, 1],
    "Fine-Tuned": [25, 8, 7],
    "Shift": [19, 8, 6]
}

# Create a DataFrame
df = pd.DataFrame(data)

# Plotting
plt.figure(figsize=(6, 4))  # Adjust size to fit in a paper

# Plot bars for Baseline and Fine-Tuned side by side
bar_width = 0.35
index = range(len(df))

plt.bar(index, df['Baseline'], width=bar_width, label='Baseline', color='skyblue')
plt.bar([i + bar_width for i in index], df['Fine-Tuned'], width=bar_width, label='Fine-Tuned', color='salmon')

# Adding labels and title
plt.xlabel('Emotion')
plt.ylabel('Percentage (%)')
plt.title('Emotion Bias Detection Results')
plt.xticks([i + bar_width / 2 for i in index], df['Emotion'])
plt.legend()

# Add shift annotations
for i in index:
    plt.text(i, df['Baseline'][i] + 1, f"+{df['Shift'][i]}", ha='center', va='bottom', fontsize=10, color='black')

plt.tight_layout()
plt.show()
```

```python
# Plotting with grid
plt.figure(figsize=(6, 4))  # Adjust size to fit in a paper

# Plot bars for Baseline and Fine-Tuned side by side
bar_width = 0.35
index = range(len(df))

plt.bar(index, df['Baseline'], width=bar_width, label='Baseline', color='skyblue')
plt.bar([i + bar_width for i in index], df['Fine-Tuned'], width=bar_width, label='Fine-Tuned', color='salmon')

# Adding labels and title
plt.xlabel('Emotion')
plt.ylabel('Percentage (%)')
plt.title('Emotion Bias Detection Results')
plt.xticks([i + bar_width / 2 for i in index], df['Emotion'])
plt.legend()

# Add shift annotations
for i in index:
    plt.text(i, df['Baseline'][i] + 1, f"+{df['Shift'][i]}", ha='center', va='bottom', fontsize=10, color='black')

# Add grid
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import numpy as np

# Data from the table
cohen_d_data = {
    "Apathy": [np.nan, 0.0634, 0.1053, 0.0100],
    "Joy": [0.0634, np.nan, 0.1771, 0.0803],
    "Fear": [0.1053, 0.1771, np.nan, 0.0920],
    "Anger": [0.0100, 0.0803, 0.0920, np.nan]
}

# Create a DataFrame
cohen_d_df = pd.DataFrame(cohen_d_data)

# Set the index to match the emotions
cohen_d_df.index = cohen_d_data.keys()

# Plotting the heatmap
plt.figure(figsize=(8, 4))
sns.heatmap(cohen_d_df, annot=True, cmap='coolwarm', cbar_kws={'label': 'Cohen\'s d Coefficient'})

plt.title('Cohen\'s d Coefficients')
plt.tight_layout()
plt.show()
```
